## 1. La Problemática de los Tipos en Python y la Solución Pydantic v2

Python es un lenguaje de **tipado dinámico y tipado fuerte**. Aunque las variables pueden cambiar de tipo en tiempo de ejecución, la falta de verificación en tiempo de compilación suele derivar en errores en producción como `TypeError`, `AttributeError` o cargas malformadas recibidas en APIs REST o mediante Web Scraping.

````python
# Ejemplo 1: Demostración de Pydantic v2 para coerción estricta de tipos y validación
from datetime import datetime
from pydantic import BaseModel, Field, field_validator

class ScrapedCommitModel(BaseModel):
    hash: str = Field(..., min_length=7, max_length=40)
    author: str = Field(..., min_length=1)
    additions: int = Field(0, ge=0)
    deletions: int = Field(0, ge=0)
    timestamp: datetime = Field(default_factory=datetime.utcnow)

    @field_validator('hash')
    @classmethod
    def validate_hash(cls, v: str) -> str:
        v_clean = v.strip().lower()
        if not v_clean.isalnum():
            raise ValueError('El hash debe ser alfanumérico')
        return v_clean

# Probar coerción automática (convierte texto "120" a entero 120 e ISO String a Datetime)
commit_raw = {
    'hash': 'A1B2C3D4E5F67890',
    'author': '@tiangolo',
    'additions': '120',
    'deletions': 5,
    'timestamp': '2026-09-03T17:30:00Z'
}

validated_commit = ScrapedCommitModel(**commit_raw)
print('Commit Validado y Coercionado con Exito:')
print(validated_commit.model_dump())
print(f'Tipo de additions: {type(validated_commit.additions)}')
print(f'Tipo de timestamp: {type(validated_commit.timestamp)}')
````

## 2. Web Scraping de GitHub con BeautifulSoup4 e HTTPX

Utilizaremos `httpx` para peticiones HTTP asíncronas con rotación de `User-Agent` (siguiendo las recomendaciones de ScraperAPI) y `bs4` para extraer selectores HTML.

````python
import asyncio
from scraper import GitHubWebScraper

async def test_scraper():
    scraper = GitHubWebScraper()
    repo_url = 'https://github.com/fastapi/fastapi'
    owner, repo_name = scraper.extract_repo_owner_name(repo_url)
    print(f'Analizando repositorio: {owner}/{repo_name}')
    
    html = await scraper.fetch_commits_html(repo_url)
    commits = scraper.parse_commits_html(html, f'{owner}/{repo_name}', limit=5)
    
    print(f'Commits extraidos ({len(commits)}):')
    for c in commits:
        print(f' - [{c.short_hash}] Autor: @{c.author_username} | Msg: {c.message[:40]}... (Padre: {c.parent_hash[:7] if c.parent_hash else "Nulo"})')

# Ejecutar test en el notebook
await test_scraper()
````
