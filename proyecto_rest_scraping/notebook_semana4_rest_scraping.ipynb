## 1. Importación de Librerías y Configuración Inicial

```python
import httpx
from bs4 import BeautifulSoup
import sqlite3
import pandas as pd
from pathlib import Path

# Verificar versiones instaladas
print(f"httpx versión: {httpx.__version__}")
    print(f"pandas versión: {pd.__version__}")
```

## 2. Consumo Estructurado de APIs REST (Open-Meteo)

```python
# Endpoint público de API REST meteorológica
URL_API = "https://api.open-meteo.com/v1/forecast"
PARAMS = {
    "latitude": -25.2637,   # Asunción, Paraguay
    "longitude": -57.5759,
    "current_weather": True
}

with httpx.Client(timeout=10.0) as client:
    response = client.get(URL_API, params=PARAMS)
    print(f"Estado HTTP Response: {response.status_code}")
    data = response.json()

# Extraer payload meteorológico actual
clima = data['current_weather']
print("Datos procesados de la API REST:")
print(f"- Temperatura: {clima['temperature']} °C")
print(f"- Velocidad del Viento: {clima['windspeed']} km/h")
print(f"- Dirección del Viento: {clima['winddirection']}°")
```

## 3. Extracción Sintáctica mediante Web Scraping (BeautifulSoup4)

```python
URL_SCRAPE = "https://news.ycombinator.com/"
HEADERS = {"User-Agent": "MecatronicaBot/1.0 (Educational Scraping Notebook)"}

with httpx.Client(headers=HEADERS, timeout=10.0) as client:
    resp = client.get(URL_SCRAPE)
    soup = BeautifulSoup(resp.text, "html.parser")

# Extracción mediante Selectores CSS (.titleline > a)
links = soup.select(".titleline > a")
titulares = []
for idx, link in enumerate(links[:5], start=1):
    titulares.append({
        "posicion": idx,
        "titulo": link.get_text(strip=True),
        "url": link.get("href", "")
    })

df_scraped = pd.DataFrame(titulares)
df_scraped
```

## 4. Persistencia en Base de Datos Relacional SQLite

```python
DB_FILE = Path("data/notebook_demo.db")
DB_FILE.parent.mkdir(parents=True, exist_ok=True)

# Crear conexión SQLite y tablas
conn = sqlite3.connect(DB_FILE)
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS registro_clima (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    temperatura REAL,
    velocidad_viento REAL,
    registrado_en TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
""")

# Inserción de la lectura obtenida de la API REST
cursor.execute("""
INSERT INTO registro_clima (temperatura, velocidad_viento)
VALUES (?, ?);
""", (clima['temperature'], clima['windspeed']))
conn.commit()
print("Lectura insertada exitosamente en SQLite!")
```

## 5. Consulta y Análisis de Datos con Pandas

```python
# Cargar tabla de SQLite directamente en DataFrame de Pandas
df_clima = pd.read_sql_query("SELECT * FROM registro_clima;", conn)
conn.close()

print("Tabla de registros almacenados:")
df_clima
```
